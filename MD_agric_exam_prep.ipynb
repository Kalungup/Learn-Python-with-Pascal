```python
def temperature_range(field_id):
    
    min_temp = MD_agric_df.loc[
        MD_agric_df['Field_ID'] == field_id,
        'Min_temperature_C'
    ].min()

    max_temp = MD_agric_df.loc[
        MD_agric_df['Field_ID'] == field_id,
        'Max_temperature_C'
    ].max()


    
    temp_range = max_temp - min_temp
    
    return temp_range

range_1458 = temperature_range(1458)
range_1895 = temperature_range(1895)
range_5443 = temperature_range(5443)

answers = (range_1458, range_1895, range_5443)
answers
```
